```python
from ucimlrepo import fetch_ucirepo
import pandas as pd
import numpy as np
from scipy.stats import skew
import seaborn as sns
import matplotlib.pyplot as plt
from pandas.api.types import is_numeric_dtype
dataset = fetch_ucirepo(id=967)

X = dataset.data.features
y = dataset.data.targets
rows = []

# ---- stats + per-column plots (your original behavior) ----
for column in X.select_dtypes(include=[np.number]).columns:
    if not is_numeric_dtype(X[column]):
        continue

    s = X[column].dropna()

    mean = s.mean()
    median = s.median()
    std = s.std()
    min_val = s.min()
    max_val = s.max()
    q1 = s.quantile(0.25)
    q3 = s.quantile(0.75)
    iqr = q3 - q1

    print(column)
    print("mean = " + str(mean))
    print("median = " + str(median))
    print("std = " + str(std))
    print("min_val = " + str(min_val))
    print("max_val = " + str(max_val))
    print("q1 = " + str(q1))
    print("q3 = " + str(q3))
    print("iqr = " + str(iqr))

    plt.figure()
    plt.boxplot(s)
    plt.title(f"Boxplot: {column}")
    plt.show()

    plt.figure()
    plt.hist(s, bins=30)
    plt.xlabel(column)
    plt.ylabel("Frequency")
    plt.title(f"Histogram: {column}")
    plt.show()

    print("\n" * 6)

    rows.append({
        "Feature": column,
        "Mean": mean,
        "Median": median,
        "Std Dev": std,
        "Min": min_val,
        "Q1": q1,
        "Q3": q3,
        "Max": max_val,
        "IQR": iqr
    })

# ---- summary table dataframe (same as chat table) ----
summary_df = pd.DataFrame(rows, columns=[
    "Feature","Mean","Median","Std Dev","Min","Q1","Q3","Max","IQR"
])

# optional: 2-decimal formatting (for display/export)
summary_df_2dp = summary_df.copy()
for c in ["Mean","Median","Std Dev","Min","Q1","Q3","Max","IQR"]:
    summary_df_2dp[c] = summary_df_2dp[c].astype(float).round(2)

print(summary_df_2dp)

# ---- bar plots for each metric column in the summary table ----
# (one bar plot per metric; 8 plots total)
metrics = ["Mean","Median","Std Dev","Min","Q1","Q3","Max","IQR"]

for m in metrics:
    plt.figure(figsize=(12, 5))
    plt.bar(summary_df["Feature"], summary_df[m])
    plt.title(f"{m} by Feature")
    plt.ylabel(m)
    plt.xticks(rotation=90)
    plt.tight_layout()
    plt.show()
```

# Distribution Properties

```python
from ucimlrepo import fetch_ucirepo
import pandas as pd
import numpy as np
from scipy.stats import skew
import seaborn as sns
import matplotlib.pyplot as plt

dataset = fetch_ucirepo(id=967)

X = dataset.data.features
y = dataset.data.targets

df = pd.concat([X, y], axis=1)
df.head()

data = df.select_dtypes(include='number').values.flatten()
data = data[~np.isnan(data)]

# Calculate skewness
skewness = skew(data, bias=False)
skewness = float(skewness)
print("Skewness:", skewness)

# KDE plot
plt.figure(figsize=(14, 6))
sns.kdeplot(data, fill=True, color='steelblue', linewidth=2)
plt.title(f"KDE Plot of Dataset\nOverall Skewness = {skewness:.3f}", fontsize=16, weight="bold")
plt.xlabel("Value", fontsize=14)
plt.ylabel("Density", fontsize=14)
plt.tight_layout()
plt.show()

# Box plot
plt.figure(figsize=(10, 4))
sns.set_theme(style="whitegrid")
sns.boxplot(x=data, color="orange", linewidth=1.5, fliersize=3)
plt.title("Box Plot of Entire Dataset", fontsize=16, weight="bold")
plt.xlabel("Value", fontsize=14)
plt.tight_layout()
plt.show()
```

# Correlation Analysis

```python
# Numeric Columns
numeric_df = df.select_dtypes(include=[np.number])
print(f"Numeric columns: {len(numeric_df.columns)}")
print(numeric_df.columns.tolist())
```

```python
# Covariance Matrix
# set_option('precision', 3)
covariance = numeric_df.cov()
print("Covariance Matrix:")
display(covariance)
```

```python
# Correlation Matrix
correlations = numeric_df.corr(method='pearson')
print("\nPearson Correlation Matrix:")
display(correlations)
```

```python
# Heatmap of Pearson Correlations
plt.figure(figsize=(20, 20))
sns.heatmap(correlations, vmin=-1, vmax=1, annot=True, fmt=".2f", cmap="YlGnBu")
plt.title("Pearson Correlation Heatmap")
plt.show()
```

```python
# Correlation strength (sorted)
corr_pairs = correlations.abs().unstack()
corr_pairs = corr_pairs[corr_pairs < 1.0]
corr_pairs = corr_pairs.sort_values(ascending=False)

print("15 strongest correlation pairs:")
print(corr_pairs.drop_duplicates().head(15))
```

```python
# Scatter plots for top correlated features
top_pairs = corr_pairs.drop_duplicates().head(5).index.tolist()

for x_col, y_col in top_pairs:
    numeric_df.plot(x=x_col, y=y_col, kind='scatter', figsize=(6,4))
    plt.title(f"Scatter Plot: {x_col} vs {y_col}")
    plt.show()
```

```python
# categorical columns
categorical_df = df.select_dtypes(include=['object', 'category'])

print(f"Categorical columns: {len(categorical_df.columns)}")
print(categorical_df.columns.tolist())
```

```python
# both only have 0/1 value, not really counts ---> we will treat them as categorical
print(categorical_df['NoOfURLRedirect'].value_counts())
print(categorical_df['NoOfSelfRedirect'].value_counts())
```

```python
categorical_df.head()
```

```python
import scipy.stats as sp
from scipy.stats import chi2_contingency
from scipy.stats import chi2

# significance level
alpha = 0.05

chi2_results = []

drop_cols = ['URL', 'Domain', 'Title', 'TLD', 'label']
categorical_df = categorical_df.drop(columns=drop_cols, errors='ignore')
cols = categorical_df.columns

for i in range(len(cols)):
  for j in range(i + 1, len(cols)):
    # observed contingency table
    observed_table = pd.crosstab(categorical_df[cols[i]], categorical_df[cols[j]])

    # chi-square test
    chi2, p, dof, expected = sp.chi2_contingency(observed_table)

    chi2_results.append({
      "A1": cols[i],
      "A2": cols[j],
      "Chi-Square Test Statistic": chi2,
      "p-value": p,
      "Degrees of Freedom": dof,
      "Decision": "Dependent" if p <= alpha else "Independent"
    })

results_df = pd.DataFrame(chi2_results)
display(results_df)
```

# Feature Range and Scale - Arrshan


```python
from ucimlrepo import fetch_ucirepo
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

dataset = fetch_ucirepo(id=967)
X = dataset.data.features
y = dataset.data.targets

# Separate numeric features (exclude binary/categorical)
binary_cols = [col for col in X.columns if set(X[col].dropna().unique()).issubset({0, 1})]
numeric_features = X.drop(columns=binary_cols).select_dtypes(include=np.number)

# Feature range summary
range_summary = pd.DataFrame({
    'Min': numeric_features.min(),
    'Max': numeric_features.max(),
    'Range': numeric_features.max() - numeric_features.min(),
    'Mean': numeric_features.mean(),
    'Std': numeric_features.std()
}).sort_values('Range', ascending=False)

display(range_summary)
```

```python
plt.figure(figsize=(14, 5))
range_summary['Range'].plot(kind='bar', color='steelblue')
plt.title('Feature Range (Max - Min) per Numeric Feature', fontsize=14, weight='bold')
plt.ylabel('Range')
plt.xlabel('Feature')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler

# Features already bounded between 0 and 1 → min-max scale
bounded_features = ['URLCharProb', 'TLDLegitimateProb', 'CharContinuationRate',
                    'LetterRatioInURL', 'DegitRatioInURL', 'SpacialCharRatioInURL', 'ObfuscationRatio']

# Remaining numeric features → z-score standardize
zscore_features = [col for col in numeric_features.columns if col not in bounded_features]

scaler_z = StandardScaler()
scaler_mm = MinMaxScaler()

X_zscore = pd.DataFrame(scaler_z.fit_transform(numeric_features[zscore_features]),
                         columns=zscore_features)
X_minmax = pd.DataFrame(scaler_mm.fit_transform(numeric_features[bounded_features]),
                         columns=bounded_features)

X_scaled = pd.concat([X_zscore, X_minmax], axis=1)
print("Scaled feature shape:", X_scaled.shape)
display(X_scaled.describe().round(3))
```

```python
sample_features = ['URLLength', 'NoOfLettersInURL', 'LargestLineLength', 'LineOfCode']

fig, axes = plt.subplots(2, len(sample_features), figsize=(18, 8))

for i, col in enumerate(sample_features):
    # Before scaling
    axes[0, i].hist(numeric_features[col], bins=50, color='salmon', edgecolor='black')
    axes[0, i].set_title(f'{col}\n(Original)', fontsize=10)
    axes[0, i].set_ylabel('Frequency')

    # After z-score scaling
    axes[1, i].hist(X_zscore[col], bins=50, color='steelblue', edgecolor='black')
    axes[1, i].set_title(f'{col}\n(Z-score Scaled)', fontsize=10)
    axes[1, i].set_ylabel('Frequency')

plt.suptitle('Feature Distributions: Before vs. After Scaling', fontsize=13, weight='bold')
plt.tight_layout()
plt.show()
```
